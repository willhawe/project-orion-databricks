# Project Orion - Workspace Smoke Test

Confirms that the Databricks workspace, Unity Catalog schemas, and Delta table writes are working.

```python
CATALOG = "dbw_orion_dev_uks_001"

SCHEMAS = [
    "orion_bronze",
    "orion_silver",
    "orion_gold",
    "orion_audit",
    "orion_config",
]

for schema in SCHEMAS:
    spark.sql(f"""
        CREATE SCHEMA IF NOT EXISTS {CATALOG}.{schema}
        COMMENT 'Project Orion schema: {schema}'
    """)
```

```python
spark.sql(f"""
CREATE OR REPLACE TABLE {CATALOG}.orion_audit.workspace_smoke_test AS
SELECT
  current_timestamp() AS test_timestamp,
  current_user() AS executed_by,
  current_catalog() AS current_catalog_name,
  '{CATALOG}' AS project_catalog,
  'Project Orion Databricks workspace is ready' AS status
""")
```

```python
display(
    spark.sql(f"""
        SELECT *
        FROM {CATALOG}.orion_audit.workspace_smoke_test
    """)
)
```
